# C++ Pointers and Memory Management

Pointers are variables that store the memory address of another variable. They are fundamental to C++ for dynamic memory management, efficient array handling, and implementing complex data structures.

## 1. Pointer Basics
A pointer is declared using the `*` operator. The type of the pointer must match the type of the variable it points to.

```cpp
int val = 42;
int* ptr = &val; // ptr now holds the address of val

std::cout << "Value: " << val << "\n";
std::cout << "Address of val: " << &val << "\n";
std::cout << "Pointer ptr stores: " << ptr << "\n";
std::cout << "Value pointed to by ptr: " << *ptr << "\n";
```

## 2. Address-of (&) and Dereference (*) Operators
- `&` (Address-of): Returns the memory address of a variable.
- `*` (Dereference): Accesses the value stored at the memory address held by the pointer.

```cpp
int a = 10;
int* p = &a;

*p = 20; // Change value of 'a' through the pointer
std::cout << "New value of a: " << a << "\n";
```

## 3. Null Pointers
A null pointer is a pointer that does not point to any memory location. In modern C++, use `nullptr` instead of `NULL` or `0`.

```cpp
int* p_null = nullptr;

if (p_null == nullptr) {
    std::cout << "Pointer is null\n";
}
```

## 4. Dynamic Memory Allocation
Dynamic memory is allocated on the **heap** using `new` and must be manually released using `delete` to avoid memory leaks.

```cpp
// Allocate a single integer on the heap
int* dyn_int = new int(100);
std::cout << "Dynamic value: " << *dyn_int << "\n";
delete dyn_int; // Free memory

// Allocate an array on the heap
int* dyn_arr = new int[5]{1, 2, 3, 4, 5};
std::cout << "Array element 0: " << dyn_arr[0] << "\n";
delete[] dyn_arr; // Free array memory
```

## 5. Smart Pointers
To avoid manual memory management and leaks, C++ provides smart pointers in the `<memory>` header.

- `std::unique_ptr`: Exclusive ownership of a resource.
- `std::shared_ptr`: Shared ownership via reference counting.

```cpp
#include <memory>

// Unique pointer
std::unique_ptr<int> u_ptr = std::make_unique<int>(50);
std::cout << "Unique ptr value: " << *u_ptr << "\n";

// Shared pointer
std::shared_ptr<int> s_ptr1 = std::make_shared<int>(100);
std::shared_ptr<int> s_ptr2 = s_ptr1; // Both point to same memory
std::cout << "Shared ptr value: " << *s_ptr2 << "\n";
```
