# C++ Functions and Scope

This notebook covers how to define functions, manage parameters, and understand where variables live (scope).

## 1. Function Basics
A function consists of a **declaration** (prototype) and a **definition** (implementation).

```cpp
// Declaration (usually in a header file)
int add(int a, int b);

// Definition
int add(int a, int b) {
    return a + b;
}

// Usage
int result = add(5, 3); // result = 8
```

## 2. Parameter Passing
C++ provides three main ways to pass data to functions.

```cpp
#include <iostream>

// Pass by Value: Creates a copy. Original is unchanged.
void byValue(int x) {
    x = 10;
}

// Pass by Reference: No copy. Original is modified.
void byReference(int &x) {
    x = 20;
}

// Pass by Pointer: Pass address. Original is modified.
void byPointer(int *x) {
    *x = 30;
}

int main() {
    int val = 5;
    
    byValue(val);     // val still 5
    byReference(val); // val becomes 20
    byPointer(&val);   // val becomes 30
    
    return 0;
}
```

## 3. Scope and Lifetime
- **Local Scope**: Variables defined inside a function/block. Destroyed when block ends.
- **Global Scope**: Variables defined outside all functions. Live for program duration.

```cpp
int globalVar = 100; // Global scope

void scopeDemo() {
    int localVar = 10; // Local scope
    std::cout << globalVar << " " << localVar << "\n";
}

// std::cout << localVar; // Error: localVar is not visible here
```

## 4. Static Variables
A `static` local variable retains its value between function calls.

```cpp
void counter() {
    static int count = 0; // Initialized once
    count++;
    std::cout << "Count: " << count << "\n";
}

// Calling counter() three times prints: 1, 2, 3
```
