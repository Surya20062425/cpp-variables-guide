# Variables in C++

This notebook provides a comprehensive guide to understanding variables in C++. Variables are the basic units of data storage in any program.

## 1. What is a Variable?

A variable is a container (a named memory location) for storing data values. In C++, you must declare the **type** of the variable before using it.

```python
// Example of a simple variable declaration
int myAge = 25; // Integer variable
```

## 2. Basic Data Types

C++ provides several built-in data types to handle different kinds of data:

| Type | Description | Size (typical) | Example |
| :--- | :--- | :--- | :--- |
| `int` | Integers (whole numbers) | 4 bytes | `int x = 10;` |
| `double` | Floating-point numbers | 8 bytes | `double pi = 3.14159;` |
| `float` | Floating-point numbers | 4 bytes | `float temp = 98.6f;` |
| `char` | Single character | 1 byte | `char grade = 'A';` |
| `string` | Sequence of characters | Variable | `string name = "Surya";` |
| `bool` | Boolean (true or false) | 1 byte | `bool isCoding = true;` |

```python
#include <iostream>
#include <string>
using namespace std;

int main() {
    int age = 20;
    double salary = 55000.50;
    char initial = 'S';
    string name = "Surya";
    bool isStudent = true;

    cout << "Name: " << name << endl;
    cout << "Age: " << age << endl;
    cout << "Initial: " << initial << endl;
    cout << "Salary: " << salary << endl;
    cout << "Student: " << isStudent << endl;

    return 0;
}
```

## 3. Variable Naming Rules

When naming variables in C++, you must follow these rules:
1. Names can contain letters, digits, and underscores (`_`).
2. Names must begin with a letter or an underscore, **not** a digit.
3. Names are **case-sensitive** (`myVar` and `myvar` are different).
4. Names cannot contain whitespaces or special characters (like `!`, `#`, `%`, etc.).
5. Names cannot be **C++ keywords** (like `int`, `while`, `return`).

## 4. Constants (`const` keyword)

If you want a variable to be read-only (its value cannot be changed after initialization), use the `const` keyword.

```python
const double PI = 3.14159;
// PI = 3.14; // This would cause a compilation error
```
