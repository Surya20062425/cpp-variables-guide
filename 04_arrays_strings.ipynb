# C++ Arrays and Strings

This notebook covers fixed-size arrays, dynamic vectors, and the difference between C-style strings and `std::string`.

## 1. Static Arrays
Fixed-size sequences of elements of the same type. Size must be known at compile time.

```cpp
int numbers[5] = {10, 20, 30, 40, 50};
numbers[0] = 100; // Update first element

// Iteration
for(int i = 0; i < 5; ++i) {
    std::cout << numbers[i] << " ";
}
```

## 2. std::vector
Dynamic arrays that can grow in size. Preferred over static arrays in most cases.

```cpp
#include <vector>
#include <algorithm>

std::vector<int> vec = {3, 1, 4, 1, 5};
vec.push_back(9); // Add to end

// Sorting
std::sort(vec.begin(), vec.end());

// Range-based for loop
for(int x : vec) {
    std::cout << x << " ";
}
```

## 3. Strings
C++ supports both C-style strings (null-terminated char arrays) and the safer `std::string` class.

```cpp
#include <string>

// C-style string
const char* oldSchool = "Hello"; 

// std::string
std::string modern = "Hello";
modern += " World"; // Concatenation

std::cout << modern << " (Length: " << modern.length() << ")";
```
