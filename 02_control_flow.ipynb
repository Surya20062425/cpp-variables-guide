# C++ Control Flow

Control flow statements allow you to control the execution flow of your program based on conditions and loops.

## 1. Conditional Statements (if, else if, else)

Used to execute a block of code only if a specified condition is true.

```cpp
#include <iostream>
using namespace std;

int main() {
    int num = 10;
    if (num > 0) {
        cout << "Positive number" << endl;
    } else if (num < 0) {
        cout << "Negative number" << endl;
    } else {
        cout << "Zero" << endl;
    }
    return 0;
}
```

## 2. Switch Statement

A more efficient way to handle multiple discrete values for a single variable.

```cpp
#include <iostream>
using namespace std;

int main() {
    int day = 3;
    switch (day) {
        case 1: cout << "Monday"; break;
        case 2: cout << "Tuesday"; break;
        case 3: cout << "Wednesday"; break;
        case 4: cout << "Thursday"; break;
        case 5: cout << "Friday"; break;
        default: cout << "Weekend";
    }
    return 0;
}
```

## 3. For Loop

Used when you know exactly how many times the loop should run.

```cpp
#include <iostream>
using namespace std;

int main() {
    for (int i = 1; i <= 5; i++) {
        cout << "Iteration: " << i << endl;
    }
    return 0;
}
```

## 4. While Loop

Repeats a block of code as long as a condition is true. Checks condition *before* execution.

```cpp
#include <iostream>
using namespace std;

int main() {
    int i = 5;
    while (i > 0) {
        cout << i << "... ";
        i--;
    }
    cout << "Liftoff!" << endl;
    return 0;
}
```

## 5. Do-While Loop

Similar to while, but checks the condition *after* execution, ensuring the block runs at least once.

```cpp
#include <iostream>
using namespace std;

int main() {
    int i = 0;
    do {
        cout << "This runs at least once!" << endl;
        i++;
    } while (i < 0); 
    return 0;
}
```
