# C++ Object-Oriented Programming (OOP) Basics

Object-Oriented Programming is a paradigm based on the concept of "objects", which can contain data (attributes) and code (methods). This approach helps in organizing complex software by mirroring real-world entities.

## 1. Classes and Objects
- **Class**: A blueprint for creating objects. It defines a data type by bundling data and methods together.
- **Object**: An instance of a class.

```cpp
#include <iostream>

class Car {
public:
    std::string brand;
    int year;

    void drive() {
        std::cout << "The " << brand << " is driving!" << std::endl;
    }
};

int main() {
    Car myCar; // Create an object of Car
    myCar.brand = "Toyota";
    myCar.year = 2022;
    myCar.drive();
    return 0;
}
```

## 2. Access Modifiers
Access modifiers define the visibility of class members:
- `public`: Members are accessible from outside the class.
- `private`: Members are accessible only within the class (default for classes).
- `protected`: Members are accessible within the class and by inherited classes.

```cpp
#include <iostream>

class Employee {
private:
    double salary; // Private: cannot be accessed directly from outside

public:
    std::string name;

    void setSalary(double s) {
        if (s > 0) salary = s;
    }

    double getSalary() {
        return salary;
    }
};

int main() {
    Employee emp;
    emp.name = "Alice"; // OK
    // emp.salary = 50000; // Error: salary is private
    emp.setSalary(50000);
    std::cout << emp.name << " earns " << emp.getSalary() << std::endl;
    return 0;
}
```

## 3. Constructors and Destructors
- **Constructor**: A special member function that is called automatically when an object is created. It typically initializes data members.
- **Destructor**: A special member function called automatically when an object is destroyed (goes out of scope or `delete` is called).

```cpp
#include <iostream>

class DatabaseConnection {
public:
    // Constructor
    DatabaseConnection() {
        std::cout << "Connecting to database..." << std::endl;
    }

    // Destructor
    ~DatabaseConnection() {
        std::cout << "Closing database connection..." << std::endl;
    }
};

int main() {
    {
        DatabaseConnection db; // Constructor called here
    } // Destructor called here as db goes out of scope
    return 0;
}
```

## 4. Encapsulation
Encapsulation is the practice of bundling data and the methods that operate on that data into a single unit (class) and restricting direct access to some of the object's components. This is achieved using `private` members and `public` getter/setter methods.

```cpp
#include <iostream>

class BankAccount {
private:
    double balance = 0;

public:
    void deposit(double amount) {
        if (amount > 0) balance += amount;
    }

    void withdraw(double amount) {
        if (amount > 0 && amount <= balance) balance -= amount;
        else std::cout << "Insufficient funds!" << std::endl;
    }

    double getBalance() const { return balance; }
};

int main() {
    BankAccount account;
    account.deposit(100);
    account.withdraw(30);
    std::cout << "Balance: " << account.getBalance() << std::endl;
    return 0;
}
```

## 5. Inheritance Basics
Inheritance allows a class (derived class) to inherit attributes and methods from another class (base class). This promotes code reuse.

```cpp
#include <iostream>

class Animal {
public:
    void eat() {
        std::cout << "Eating..." << std::endl;
    }
};

class Dog : public Animal {
public:
    void bark() {
        std::cout << "Woof! Woof!" << std::endl;
    }
};

int main() {
    Dog myDog;
    myDog.eat();  // Inherited from Animal
    myDog.bark(); // Dog's own method
    return 0;
}
```
